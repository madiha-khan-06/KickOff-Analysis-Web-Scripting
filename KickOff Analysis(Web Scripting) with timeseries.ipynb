## Read .csv from Multiple URLs with Pandas

#### Link: root + season + league

```python
# link structure
"https://www.football-data.co.uk/mmz4281/" + "2122" + "/" + "E0" + ".csv"
```

```python
# create a root variable
root = "https://www.football-data.co.uk/mmz4281/"
```

### Multiple leagues

```python
# creating list of leagues
leagues = ['E0','E2', 'E3']

frames = []
# looping through leagues, read multiple csv and append it into a list
for league in leagues:
    df = pd.read_csv(root + "2122" + "/"  + league + ".csv")
    frames.append(df)
```

```python
# length of frames
len(frames)
```

```python
# show 1st, 2nd and 3rd element
frames[2].head(3)
```

### Multiple Seasons

```python
import pandas as pd
import time

frames = []

for league in leagues:
    for season in range(15, 25):
        url = f"{root}{season}{season+1}/{league}.csv"
        print("Loading:", url)

        try:
            df = pd.read_csv(url)

            # avoid duplicate column error
            if 'season' not in df.columns:
                df.insert(1, 'season', season)

            frames.append(df)
            time.sleep(2)  # IMPORTANT

        except Exception as e:
            print(f"Skipped {url} -> {e}")
```

```python
# total frames (10 seasons x 4 leagues)
len(frames)
```

```python
# show 1st, and last element (new season column was created)
frames[1]
```

```python
frames[-22]
```

### Organize All The Data in a Dictionary

```python
dict_countries = {
              'Spanish La Liga':'SP1', 'Spanish Segunda Division':'SP2',
              'German Bundesliga':'D1', 'German Bundesliga 2':'D2', 
              'English Premier League':'E0', 'English League 1':'E2', 'English League 2':'E3',
             }
```

```python
dict_countries['German Bundesliga 2']
```

```python
import time
import pandas as pd

dict_historical_data = {}

for league_name, league_code in dict_countries.items():
    print(f"\nProcessing league: {league_name}")
    frames = []

    for season in range(15, 25):
        url = f"{root}{season}{season+1}/{league_code}.csv"
        print("  Loading:", url)

        try:
            df = pd.read_csv(url)

            # prevent duplicate column error
            if 'season' not in df.columns:
                df.insert(1, 'season', season)

            frames.append(df)
            time.sleep(2)  # IMPORTANT

        except Exception as e:
            print(f"  Skipped season {season}: {e}")

    if frames:  # only concat if data exists
        dict_historical_data[league_name] = pd.concat(frames, ignore_index=True)
```

```python
# show dataframe inside dict_historical_data
dict_historical_data['German Bundesliga 2']
```

## DATA VISUALIZATION

### 1. Matches per Season (Premier League)

```python
import matplotlib.pyplot as plt

pl = dict_historical_data['English Premier League']


pl['season'].value_counts().sort_index().plot(kind='bar')
plt.title('Premier League Matches per Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')
plt.show()
```

### 2. Average Goals per Match by Season

```python
pl['total_goals'] = pl['FTHG'] + pl['FTAG']

pl.groupby('season')['total_goals'].mean().plot(marker='o')
plt.title('Average Goals per Match by Season')
plt.xlabel('Season')
plt.ylabel('Goals')
plt.show()
```

### 3. Home vs Away Goals Comparison

```python
pl[['FTHG', 'FTAG']].mean().plot(kind='bar')
plt.title('Average Home vs Away Goals')
plt.ylabel('Goals')
plt.show()
```

### 4. Top 10 team by Home Win

```python
pl[['FTHG', 'FTAG']].mean().plot(kind='bar')
plt.title('Average Home vs Away Goals')
plt.ylabel('Goals')
plt.show()
```

### 5.Average Goals per match by League

```python
league_goals = {}

for league, df in dict_historical_data.items():
    league_goals[league] = (df['FTHG'] + df['FTAG']).mean()

pd.Series(league_goals).plot(kind='bar')
plt.title('Average Goals per Match by League')
plt.ylabel('Goals')
plt.show()
```

### 6.Goal Distribution

```python
pl = dict_historical_data['English Premier League']
pl['total_goals'] = pl['FTHG']
pl['FTAG']


pl['total_goals'].hist(bins=20)
plt.title('Goals Distribution')
plt.xlabel('Goals')
plt.ylabel('Frequency')
plt.show()
```

### 7.Best Defensive Team

```python
best_def = pl.groupby('AwayTeam')['FTAG'].sum().sort_values().head(10)
best_def.plot(kind='bar')
plt.title('Best Defensive Teams')
plt.show()
```

### 8.Home Team Advantage

```python
home_win = (pl['FTR'] == 'H').mean()
away_win = (pl['FTR'] == 'A').mean()

print("Home Win %:", home_win)
print("Away Win %:", away_win)
```

### 9.Top Scoring Teams

```python
top_scoring = pl.groupby('HomeTeam')['FTHG'].sum().sort_values(ascending=False).head(10)
top_scoring.plot(kind='bar')
plt.title('Top Scoring Teams')
plt.show()
```

### 10.Correlation Heatmap

```python
import seaborn as sns
```

```python
corr = pl[['FTHG', 'FTAG', 'HTHG', 'HTAG']].corr()
sns.heatmap(corr, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```
